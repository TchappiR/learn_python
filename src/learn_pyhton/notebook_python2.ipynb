## 1- P.O.O

```python
class Vehicule:
    
    def __init__(self, marque : str, vitesse_max : float):
        self._marque = marque
        self.__vitesse_max = vitesse_max
    
    def __str__(self) -> str:
        return f"Vehicule de marque : {self._marque}"
    
    def __eq__(self, other_vehicule : "Vehicule") -> bool:
        return self._marque.lower() == other_vehicule._marque.lower() and self.__vitesse_max == other_vehicule.__vitesse_max

    def deplacer(self):
        return f"Le vehicule {self._marque} se déplace ..."
```

```python
class Voiture(Vehicule):
    def __init__(self, marque : str, vitesse_max : float, portes : int):
        super().__init__(marque, vitesse_max)
        self.portes = portes

    def deplacer(self):
        return f"Le voiture {self._marque} se route à {self.portes} portes ..."

class Moto(Vehicule):
    def deplacer(self):
        return f"La moto {self._marque} fonce ..."
```

```python
vehicules = [
    Voiture("Peugeot", 180.0, 4),
    Moto("Honda", 200.0)
]
```

```python
for v in vehicules:
    print(v.deplacer())
```

```python
v1 = Voiture("Toyota", 150.0, 4)
v2 = Voiture("Toyota", 150.0, 5)
```

```python
print(v1 == v2)  
```

## 3- Gestion Erreur

```python
class SoldeInsuffisant(Exception):
    pass


class CompteBancaire:
    def __init__(self, solde : float):
        self.solde = solde
    
    def retrait(self, montant : float):
        if montant > self.solde:
            raise SoldeInsuffisant("Le montant que vous voulez retiré est inferieur à votre solde.")    
        self.solde -= montant
```

```python
name = input("Entrez votre nom : ")
solde = int(input("Quel est votre solde ? "))
montant = int(input("Combien voulez vous retirer ? "))

obj_cpt1 = CompteBancaire(solde)
print(f"Votre solde initiale est de : {obj_cpt1.solde}")
try:
    obj_cpt1.retrait(montant)
    print(f"Retrait effectué")
except SoldeInsuffisant as e:
    print(f"Vous ne pouvez pas effectuer de retrait car : {e}")
finally:
    print(f"{name.capitalize()}, votre solde restant est de : {obj_cpt1.solde}€\nA bientôt")
```

Dans cet exercice, vous allez implémenter une fonction de traitement de données qui gère les exceptions potentielles. Vous devez :<br>

1- Créer une fonction traiter_donnees(data) qui prend une liste de données.<br>
2- Utiliser try et except pour gérer les exceptions suivantes :<br>
    ValueError si les données ne sont pas numériques.<br>
    TypeError si les données ne sont pas itérables.<br>
3- Utiliser assert pour vérifier que la liste n’est pas vide.<br>
4- Lever une exception personnalisée si les données contiennent des valeurs négatives.

```python
class NegativeValorError(Exception):
    pass

def traiter_donnees(data: list):
    try:
        assert len(data) > 0, "La liste ne doit pas être vide"
        for val in data:
            if not isinstance(val, (int, float)):
                raise ValueError(f"Erreur detectée : {val} n'est pas numerique")
            if val < 0:
                raise NegativeValorError(f"Erreur detectée : {val} est un négatif")
    except AssertionError as e:
        print(e)
    except ValueError as e:
        print(e)
    except NegativeValorError as e:
        print(e)
    except TypeError as e:
        print(f"Vous devez fournir une liste de valeurs numerique")
    else:
        print("Données enregistrées avec succès")
```

```python
valo = [12, 4]
traiter_donnees(valo)
```

## 4- Recursivité

```python
class NegativeValueError(Exception):
    pass

def factorielle(number : int) -> int:
    if not isinstance(number, int) or number < 0:
        raise TypeError(f"Impossible de calculer {number}!")
    if number <= 1:
        return 1
    return number * factorielle(number - 1)
```

```python
number = 5
try:
    resultat = factorielle(number)
except NegativeValueError as e:
    print(e)
except TypeError as e:
    print(e)
else: 
    print(f"{number}! = {resultat}")
```

```python
def fibonacci(n):
    if n <= 1:
        return n   
    resultat = fibonacci(n-1) + fibonacci(n-2)
    return resultat
```

```python
fibonacci(50)
```

```python
def fibonacci_memo(n):
    cache = {}
    def fib(n):
        if n in cache:
            return cache[n]
        if n <= 1:
            return n   
        cache[n] = fib(n-1) + fib(n-2)
        return cache[n]
    return fib(n)
```

```python
def fib_memo(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib_memo(n-1, memo) + fib_memo(n-2, memo)
    return memo[n]
    
```

```python
fib_memo(300)
```

```python
fibonacci_memo(300)
```

```python
transactions = [
    ("2023-01-15", "Électronique", "Smartphone", 599.99, "Premium"),
    ("2023-01-15", "Vêtements", "T-shirt", 19.99, "Standard"),
    ("2023-01-16", "Électronique", "Laptop", 899.99, "Premium"),
    ("2023-01-16", "Électronique", "Smartphone", 599.99, "Standard"),
    ("2023-01-17", "Alimentation", "Café", 4.50, "Standard"),
    ("2023-01-17", "Vêtements", "Jean", 49.99, "Premium"),
    ("2023-01-18", "Électronique", "Smartphone", 599.99, "Premium"),
    ("2023-01-18", "Alimentation", "Thé", 3.25, "Standard"),
]
```

```python
from collections import Counter, defaultdict, OrderedDict
```

```python
# 1- Enregistrement par categories
Counter([transaction[1] for transaction in transactions])
```

```python
# 2- Les 2 produits les plus vendus
Counter([transaction[2] for transaction in transactions]).most_common(2)
```

```python
# 3- Repartitions des types de clients
Counter(transaction[4] for transaction in transactions)
```

```python
# 4- Regrouper les transactions par date:
transations_par_date = defaultdict(list)

for date, categorie, produit, prix, status in transactions:
    transations_par_date[date].append((categorie, produit, prix, status))

# -- Affichage
for key, val in transations_par_date.items():
    print(f"{key} : {val}")
```

```python
# 5- CA par categorie
transactions_par_categorie = defaultdict(list)

for _, categorie, _, montant, _ in transactions:
    transactions_par_categorie[categorie].append(montant)

print("=== Etat des recetets par categories de produit ===")
for key, val in transactions_par_categorie.items():
    print(f"{key} : {sum(val)}€")
```

```python
# 6- Liste de produits par statut client
achats_par_status = defaultdict(list)

for _, _, produit, _, status in transactions:
    achats_par_status[status].append(produit)

for key, val in achats_par_status.items():
    print(f"{key} : {val}")
```
