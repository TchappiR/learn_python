### Exercice1 numpy

```python
from skimage import data, color
import matplotlib.pyplot as plt
image = data.retina()
face = color.rgb2gray(image)
plt.imshow(face) 
plt.show()
```

```python
face*255
```

```python
face.shape
```

```python
lines = face.shape[0]
colonnes = face.shape[1]
```

```python
# 1- Zoom d'un quart
face_zoom = face[lines//4:-lines//4, colonnes//4:-colonnes//4]
face_zoom.shape
```

```python
plt.imshow(face_zoom)
plt.show()
```

```python
# 2- Renforcer les pixels
face_zoom[face_zoom < 50/255] = 0
face_zoom[face_zoom > 200/255] = 255
```

```python
plt.imshow(face_zoom)
plt.show
```

### Exercice 2 : Numpy

```python
import numpy as np

def initialisation(m : int, n : int):
    np.random.seed(0)
    matrix = np.random.randn(m, n)
    result = np.concatenate((matrix, np.ones(shape=(m, 1))), axis=1)
    return result
```

```python
initialisation(3, 4)
```
