### Exercice1 numpy

```python
from skimage import data, color
import matplotlib.pyplot as plt
image = data.retina()
face = color.rgb2gray(image)
plt.imshow(face) 
plt.show()
```

```python
face*255
```

```python
face.shape
```

```python
lines = face.shape[0]
colonnes = face.shape[1]
```

```python
# 1- Zoom d'un quart
face_zoom = face[(3*lines)//8:(5*lines)//8, (3*colonnes)//8:(5*colonnes)//8]
face_zoom.shape
```

```python
plt.imshow(face_zoom)
plt.show()
```

```python
# 2- Renforcer les pixels
face_zoom[face_zoom < 50/255] = 0
face_zoom[face_zoom > 200/255] = 255
```

```python
plt.imshow(face_zoom)
plt.show
```
